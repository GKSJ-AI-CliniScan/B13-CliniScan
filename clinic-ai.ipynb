```python
import os
import pydicom
import cv2
import numpy as np
from tqdm import tqdm

INPUT_DIR = "/kaggle/input/vinbigdata-chest-xray-abnormalities-detection/train"
OUTPUT_DIR = "/kaggle/working/images"

os.makedirs(OUTPUT_DIR, exist_ok=True)

for file in tqdm(os.listdir(INPUT_DIR)):
    if not file.endswith(".dicom"):
        continue

    png_path = os.path.join(OUTPUT_DIR, file.replace(".dicom", ".png"))

    if os.path.exists(png_path):
        continue

    dcm = pydicom.dcmread(os.path.join(INPUT_DIR, file))
    img = dcm.pixel_array.astype(np.float32)

    # Normalize
    img = (img - img.min()) / (img.max() - img.min())
    img = (img * 255).astype(np.uint8)

    # Resize
    img = cv2.resize(img, (224, 224))

    cv2.imwrite(
        os.path.join(OUTPUT_DIR, file.replace(".dicom", ".png")),
        img
    )
```
