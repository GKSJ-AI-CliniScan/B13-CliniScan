```python
import os
import random
import shutil

BASE_PATH = r"D:\Module_2\dataset"
images_path = os.path.join(BASE_PATH, "images")
labels_path = os.path.join(BASE_PATH, "labels")

for folder in ["train", "val", "test"]:
    os.makedirs(os.path.join(images_path, folder), exist_ok=True)
    os.makedirs(os.path.join(labels_path, folder), exist_ok=True)

images = [f for f in os.listdir(images_path) if f.endswith(".png")]
random.shuffle(images)

train_split = int(0.7 * len(images))
val_split = int(0.9 * len(images))

train_images = images[:train_split]
val_images = images[train_split:val_split]
test_images = images[val_split:]

def move_files(image_list, split_type):
    for img in image_list:
        label = img.replace(".png", ".txt")
        shutil.move(os.path.join(images_path, img),
                    os.path.join(images_path, split_type, img))
        shutil.move(os.path.join(labels_path, label),
                    os.path.join(labels_path, split_type, label))

move_files(train_images, "train")
move_files(val_images, "val")
move_files(test_images, "test")

print("Split Completed")
print("Train:", len(train_images))
print("Val:", len(val_images))
print("Test:", len(test_images))
```
