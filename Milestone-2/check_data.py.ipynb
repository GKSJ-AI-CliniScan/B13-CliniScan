```python
import os
import random
import cv2
import matplotlib.pyplot as plt

BASE_PATH = r"D:\Module_2\dataset"

images_path = os.path.join(BASE_PATH, "images", "train")
labels_path = os.path.join(BASE_PATH, "labels", "train")

images = os.listdir(images_path)
sample_images = random.sample(images, 10)

for img_name in sample_images:
    img_path = os.path.join(images_path, img_name)
    label_path = os.path.join(labels_path, img_name.replace(".png", ".txt"))

    img = cv2.imread(img_path)
    h, w, _ = img.shape

    with open(label_path, "r") as f:
        lines = f.readlines()

    for line in lines:
        cls, x, y, bw, bh = map(float, line.split())
        x1 = int((x - bw/2) * w)
        y1 = int((y - bh/2) * h)
        x2 = int((x + bw/2) * w)
        y2 = int((y + bh/2) * h)
        cv2.rectangle(img, (x1,y1), (x2,y2), (0,255,0), 2)

    plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    plt.show()
```
