```python
import os

label_dirs = [
    "dataset/kaggle/working/dataset/labels/train",
    "dataset/kaggle/working/dataset/labels/val",
    "dataset/kaggle/working/dataset/labels/test"
]

errors = []

for label_dir in label_dirs:
    for file in os.listdir(label_dir):
        if file.endswith(".txt"):
            path = os.path.join(label_dir, file)

            with open(path, "r") as f:
                lines = f.readlines()

            for i, line in enumerate(lines):
                parts = line.strip().split()

                if len(parts) != 5:
                    errors.append((file, i+1, line.strip()))

                else:
                    try:
                        class_id = int(parts[0])
                        coords = list(map(float, parts[1:]))

                        if not all(0 <= c <= 1 for c in coords):
                            errors.append((file, i+1, line.strip()))

                    except:
                        errors.append((file, i+1, line.strip()))

if len(errors) == 0:
    print("✅ All labels are in correct YOLO format")
else:
    print("❌ Format errors found:")
    for e in errors[:10]:
        print(e)
```

```python
from ultralytics import YOLO
import torch
import os

print("="*60)
print("YOLOV8 OBJECT DETECTION TRAINING - 25 EPOCHS")
print("="*60)
print()

# Fix dataset.yaml encoding issue first
print("🔧 Preparing dataset.yaml...")
yaml_content = """path: dataset/kaggle/working/dataset

train: images/train
val: images/val

nc: 4

names:
  0: opacity
  1: consolidation
  2: fibrosis
  3: mass
"""

# Write with proper UTF-8 encoding (no BOM)
with open("dataset.yaml", "w", encoding="utf-8") as f:
    f.write(yaml_content)
print("✓ dataset.yaml fixed and ready")
print()

# Check GPU availability
print("📊 SYSTEM INFO:")
print(f"  GPU Available: {torch.cuda.is_available()}")
if torch.cuda.is_available():
    print(f"  GPU: {torch.cuda.get_device_name(0)}")
    device = 0
else:
    print(f"  Using CPU (slower but will still train)")
    device = "cpu"

print()

# Load pretrained model
print("📥 Loading YOLOv8s pretrained weights...")
model = YOLO("yolov8s.pt")

print("✓ Model loaded successfully")
print()

# Verify dataset config
print("📁 DATASET CONFIGURATION:")
try:
    with open("dataset.yaml", "r", encoding="utf-8") as f:
        yaml_text = f.read()
        print(yaml_text)
except Exception as e:
    print(f"❌ ERROR reading dataset.yaml: {e}")

print()
print("🔄 STARTING TRAINING...")
print("-" * 60)

try:
    # Train model with 25 epochs (as requested)
    results = model.train(
        data="dataset.yaml",
        epochs=25,
        imgsz=640,
        batch=4,
        device=device,
        patience=3,
        save=True,
        verbose=False,
        workers=0,
        plots=True
    )
    
    print("-" * 60)
    print()
    print("✅ TRAINING COMPLETED SUCCESSFULLY!")
    print()
    print("📊 RESULTS SAVED TO:")
    print("  - Weights: runs/detect/train/weights/")
    print("  - Metrics: runs/detect/train/results.csv")
    print("  - Plots:   runs/detect/train/results.png")
    
except KeyboardInterrupt:
    print("\n"+"⚠" * 30)
    print("❌ TRAINING WAS INTERRUPTED!")
    print("⚠" * 30)
    print("\nTo continue training:")
    print("1. Run this cell again (will resume from checkpoint)")
    print("2. DO NOT interrupt the cell execution (can take hours)")
    print()
    
except Exception as e:
    print(f"\n❌ ERROR: {str(e)}")
    print("\nTroubleshooting:")
    print("1. Check that dataset.yaml paths are correct")
    print("2. Verify images and labels exist in dataset folders")
    print("3. Try reducing batch size in the training parameters")
```

```python
import os
import cv2
from ultralytics import YOLO
import matplotlib.pyplot as plt

print("=== TESTING MODEL ON SAMPLE IMAGES ===\n")

# Check if trained model exists
model_path = "runs/detect/train/weights/best.pt"
if not os.path.exists(model_path):
    print("❌ ERROR: Trained model not found!")
    print(f"Expected path: {model_path}")
    print("\n⚠ Please run the training cell first to generate the model.")
    print("Make sure to let it complete all 25 epochs without interruption.\n")
else:
    # Load trained model
    model = YOLO(model_path)

    # Get sample test images
    test_img_dir = "dataset/kaggle/working/dataset/images/test"
    test_images = [f for f in os.listdir(test_img_dir) if f.endswith(('.jpg', '.png', '.jpeg'))][:5]

    print(f"Testing on {len(test_images)} sample images...\n")

    # Run inference on sample images
    for idx, img_name in enumerate(test_images, 1):
        img_path = os.path.join(test_img_dir, img_name)
        
        try:
            # Perform inference
            results = model.predict(source=img_path, conf=0.5, verbose=False)
            
            # Display results
            print(f"Image {idx}: {img_name}")
            print(f"  Detections: {len(results[0].boxes)}")
            
            if len(results[0].boxes) > 0:
                for box in results[0].boxes:
                    class_id = int(box.cls)
                    confidence = float(box.conf)
                    class_names = {0: 'opacity', 1: 'consolidation', 2: 'fibrosis', 3: 'mass'}
                    print(f"    - {class_names[class_id]}: {confidence:.2f}")
            else:
                print(f"    - No abnormalities detected")
            print()
        except Exception as e:
            print(f"Error processing {img_name}: {e}\n")

    print("✅ Inference test completed!")
```

```python
import os
import cv2
import matplotlib.pyplot as plt
from ultralytics import YOLO

print("=== VISUALIZING DETECTION RESULTS ===\n")

# Check if trained model exists
model_path = "runs/detect/train/weights/best.pt"
if not os.path.exists(model_path):
    print("❌ ERROR: Trained model not found!")
    print(f"Expected path: {model_path}")
    print("\n⚠ Please run the training cell first to generate the model.")
    print("Make sure to let it complete all 25 epochs without interruption.\n")
else:
    # Load trained model
    model = YOLO(model_path)

    # Get sample test images
    test_img_dir = "dataset/kaggle/working/dataset/images/test"
    test_images = [f for f in os.listdir(test_img_dir) if f.endswith(('.jpg', '.png', '.jpeg'))][:3]

    if len(test_images) == 0:
        print("❌ No test images found!")
    else:
        # Create figure for visualization
        fig, axes = plt.subplots(1, len(test_images), figsize=(15, 5))
        if len(test_images) == 1:
            axes = [axes]

        for idx, img_name in enumerate(test_images):
            img_path = os.path.join(test_img_dir, img_name)
            
            try:
                # Perform inference
                results = model.predict(source=img_path, conf=0.5, verbose=False)
                
                # Get annotated image
                annotated_img = results[0].plot()
                
                # Convert BGR to RGB for display
                annotated_img_rgb = cv2.cvtColor(annotated_img, cv2.COLOR_BGR2RGB)
                
                # Display
                axes[idx].imshow(annotated_img_rgb)
                axes[idx].set_title(f"{img_name}\nDetections: {len(results[0].boxes)}")
                axes[idx].axis('off')
            except Exception as e:
                axes[idx].text(0.5, 0.5, f"Error: {str(e)[:50]}", ha='center', va='center')
                axes[idx].axis('off')

        plt.tight_layout()
        plt.savefig('detection_results.png', dpi=100, bbox_inches='tight')
        print("✅ Detection visualization saved to 'detection_results.png'")
        plt.show()
```
