```python
import os
import pydicom
import cv2
import numpy as np

# INPUT: DICOM folder from competition dataset
input_path = '/kaggle/input/vinbigdata-chest-xray-abnormalities-detection/train'

# OUTPUT: PNG folder
output_path = '/kaggle/working/png_images'
os.makedirs(output_path, exist_ok=True)

files = os.listdir(input_path)[:50]   # convert only 50 images

for file in files:
    dicom_path = os.path.join(input_path, file)
    
    ds = pydicom.dcmread(dicom_path)
    img = ds.pixel_array

    # normalize to 0–255
    img = cv2.normalize(img, None, 0, 255, cv2.NORM_MINMAX)
    img = np.uint8(img)

    png_name = file.replace('.dicom', '.png')
    save_path = os.path.join(output_path, png_name)

    cv2.imwrite(save_path, img)

print("DICOM to PNG conversion done ✅")
```
